# Soal Nomor 1 — Curve Fitting (Regresi Polinomial)

**Masalah:** Data produksi gudang (bi-weekly, Jan 2018 - Des 2023) perlu dimodelkan secara matematis agar bisa diprediksi perilakunya.

**Tujuan:** Mencari model polinomial derajat 3 yang paling sesuai dengan data.

**Metode:** *Least Squares* via `np.polyfit`

**Alasan:** Polinomial derajat 3 dipilih karena data menunjukkan tren non-linear namun tidak terlalu kompleks. Derajat lebih tinggi berisiko overfitting. Metode Least Squares meminimalkan jumlah kuadrat error antara data aktual dan model, sehingga menghasilkan koefisien yang paling optimal.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_excel('aol_data.xlsx')
x = np.arange(1, 145)
y = data.values[0]
```

```python
estimasi = np.polyfit(x, y, 3)
model_polinomial = np.poly1d(estimasi)
prediksi = model_polinomial(x)

print(estimasi)
```

```python
plt.plot(x, y, 'b.', label='Data')
plt.plot(x, prediksi, 'r-', label='Model polinomial')
plt.xlabel('Waktu (bi-weekly)')
plt.ylabel('Jumlah Produksi')
plt.legend()
plt.show()
```
